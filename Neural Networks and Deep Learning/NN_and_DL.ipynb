# Vectorization

```python
import time
import numpy as np
```

```python
N = 10000000

a = np.random.rand(N)
b = np.random.rand(N)
```

### For Loop


```python
c = 0
tic = time.time()

for i in range(N):
    c += a[i] * b[i]
    
toc = time.time()

print(c)
print(f"For Loop: {1000 * (toc - tic)}ms")
```

### Vectorized


```python
tic = time.time()
c = np.dot(a, b)
toc = time.time()

print(c)
print(f"Vectorized: {1000 * (toc - tic)}ms")
```

# Broadcasting


```python
A = np.array(
    [
        [56.0, 0.0, 4.4, 68.0],
        [1.2, 104.0, 52.0, 8.0],
        [1.8, 135.0, 99.0, 0.9]
    ]
)

print(A)
```

```python
cal = A.sum(axis=0)
print(cal)
```

```python
percentage = 100 * A / cal
print(percentage)
```
